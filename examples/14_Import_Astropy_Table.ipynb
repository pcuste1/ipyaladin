# Import Astropy Table
This notebook shows how you can import an astropy table as a selection region 

```
from ipyaladin import Aladin

from astropy.coordinates import SkyCoord
from astroquery.mast import Catalogs
```

```
%load_ext autoreload
%autoreload 2
```

```
aladin = Aladin(
    survey="SDSS9 colored",
    show_coo_grid=True,
    target="TRAPPIST-1",
    coo_frame="icrs",
    fov=0.05,
    height=400,
    samp=True,
)
aladin
```

### Add a custom catalog
load a custom catalog that we want to select

```
target_name = "TRAPPIST-1"
catalog_data = Catalogs.query_region(
    coordinates=SkyCoord.from_name(target_name),
    radius=0.01,  # [deg]
    catalog="Panstarrs",
)
catalog_data.rename_columns(["raMean", "decMean"], ["ra", "dec"])


aladin.add_table(
    catalog_data, name="test-table", color="lime", shape="circle", source_size=10
)
```

Once the catalog is loaded, you can select all of the sources that match on the provided keys. 

```
aladin.select_table(catalog_data, keys=["objID"])
```
