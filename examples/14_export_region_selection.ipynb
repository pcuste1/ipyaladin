```
from ipyaladin import Aladin

from astropy.coordinates import SkyCoord
from astroquery.mast import Catalogs
```

```
target_name = "TRAPPIST-1"
catalog_data = Catalogs.query_region(
    coordinates=SkyCoord.from_name(target_name),
    radius=0.01,  # [deg]
    catalog="Panstarrs",
)
catalog_data.rename_columns(["raMean", "decMean"], ["ra", "dec"])
```

```
aladin = Aladin(
    survey="SDSS9 colored",
    show_coo_grid=True,
    target="TRAPPIST-1",
    coo_frame="icrs",
    fov=0.05,
    height=400,
    samp=True,
)
aladin
```

```
aladin.add_table(
    catalog_data, name="test-table", color="lime", shape="circle", source_size=10
)
```

```
astropy_region = aladin.selected_region

if astropy_region:
    aladin.add_graphic_overlay_from_region(astropy_region)
else:
    print("no region selected")
```
